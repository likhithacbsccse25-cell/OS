```python
%%writefile processCreation.sh
#!/bin/bash

echo "Parent Process ID: $$"



#Creating first child process
(
    echo "Child Process ID: $$, Parent Process ID: $PPID"
    sleep 5
) &

#Creating second child process
(
    echo "Child1 Process ID1: $$, Parent1 Process ID1: $PPID"
    sleep 5
) &

#Display the process tree
sleep 1 #wait to ensure child processes start

echo -e "\nProcess Tree:"
ps -o pid,ppid,cmd --forest | grep -E "bash|sleep" | grep -v "grep"
```
